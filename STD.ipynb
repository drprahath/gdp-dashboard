```python
%%writefile app.py
import streamlit as st

# -----------------------------------------------------------------------------
# PAGE CONFIGURATION
# -----------------------------------------------------------------------------
st.set_page_config(
    page_title="STI Self-Assessment",
    page_icon="🛡️",
    layout="wide",
    initial_sidebar_state="collapsed",
)

# -----------------------------------------------------------------------------
# CUSTOM CSS (Purple & Yellow Theme matching Reference Images)
# -----------------------------------------------------------------------------
st.markdown(
    """
    <style>
    /* Global Background & Font */
    .stApp {
        background-color: #F7F5F8;
        font-family: 'Georgia', 'Segoe UI', sans-serif;
    }

    /* Top Header Banner */
    .top-banner {
        background-color: #8E3A80;
        color: white;
        padding: 22px 20px 16px 20px;
        text-align: center;
        border-bottom: 4px solid #722A66;
        margin-bottom: 0px;
        border-radius: 8px 8px 0 0;
    }
    .banner-title-row {
        display: flex;
        align-items: center;
        justify-content: center;
        gap: 15px;
    }
    .yellow-check {
        background-color: #F7E859;
        color: #8E3A80;
        font-weight: bold;
        border-radius: 50%;
        width: 44px;
        height: 44px;
        display: inline-flex;
        align-items: center;
        justify-content: center;
        font-size: 24px;
    }
    .banner-title {
        font-size: 38px;
        font-weight: 500;
        margin: 0;
        font-family: 'Segoe UI', sans-serif;
    }
    .banner-subtitle {
        color: #F7E859;
        font-size: 20px;
        margin-top: 8px;
        font-family: 'Segoe UI', sans-serif;
    }

    /* Step Indicator Bar */
    .step-bar {
        background-color: #FFFFFF;
        border-bottom: 1px solid #E2DCE5;
        padding: 0px 24px;
        display: flex;
        align-items: center;
    }
    .step-tab {
        background-color: #F5EEF4;
        color: #8E3A80;
        font-weight: bold;
        font-size: 20px;
        padding: 12px 28px;
        display: inline-block;
        border-bottom: 3px solid #8E3A80;
    }

    /* Main Assessment Card */
    .assessment-card {
        background-color: #F6F6F6;
        border: 1px solid #E5E5E5;
        padding: 40px 25px;
        min-height: 360px;
    }

    /* Left Column Question Styling */
    .question-box {
        text-align: center;
        padding: 20px;
        border-right: 1px solid #DCDCDC;
        min-height: 280px;
        display: flex;
        flex-direction: column;
        justify-content: center;
        align-items: center;
    }
    .question-text {
        font-size: 26px;
        color: #111111;
        margin-top: 20px;
        font-weight: 500;
        line-height: 1.35;
    }

    /* Radio Button Customizations */
    div[role="radiogroup"] > label {
        padding: 10px 14px !important;
        font-size: 19px !important;
    }
    div[role="radiogroup"] p {
        font-size: 19px !important;
        color: #111111 !important;
    }

    /* Purple Rounded Buttons */
    div.stButton > button {
        background-color: #8E3A80 !important;
        color: white !important;
        border-radius: 30px !important;
        padding: 10px 34px !important;
        font-size: 18px !important;
        font-weight: 500 !important;
        border: none !important;
        transition: 0.2s ease-in-out;
        width: 100%;
    }
    div.stButton > button:hover {
        background-color: #6E2862 !important;
        color: #F7E859 !important;
    }

    /* Footer Note */
    .footer-note {
        font-size: 14px;
        font-style: italic;
        color: #333333;
        margin-top: 18px;
    }
    .initiative-badge {
        font-size: 14px;
        color: #444444;
        font-weight: 600;
        margin-top: 12px;
    }
    </style>
    """,
    unsafe_allow_html=True,
)

# -----------------------------------------------------------------------------
# SVG ILLUSTRATIONS (Purple & Yellow Palette, No Character)
# -----------------------------------------------------------------------------
SVGS = {
    "shield": """
    <svg width="220" height="220" viewBox="0 0 200 200" xmlns="http://www.w3.org/2000/svg">
        <circle cx="100" cy="100" r="90" fill="#F7E859" stroke="#8E3A80" stroke-width="8"/>
        <path d="M100 35 L150 55 V100 C150 135 128 162 100 175 C72 162 50 135 50 100 V55 L100 35 Z" fill="#8E3A80"/>
        <path d="M92 75 H108 V92 H125 V108 H108 V125 H92 V108 H75 V92 H92 V75 Z" fill="#F7E859"/>
    </svg>
    """,
    "age": """
    <svg width="190" height="140" viewBox="0 0 220 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="30" cy="82" r="9" fill="#D4DF38"/>
        <rect x="20" y="94" width="20" height="35" rx="4" fill="#D4DF38"/>
        <circle cx="75" cy="65" r="11" fill="#8E3A80"/>
        <rect x="62" y="79" width="26" height="50" rx="4" fill="#8E3A80"/>
        <circle cx="125" cy="48" r="13" fill="#D4DF38"/>
        <rect x="110" y="64" width="30" height="65" rx="5" fill="#D4DF38"/>
        <circle cx="178" cy="30" r="15" fill="#8E3A80"/>
        <rect x="160" y="48" width="36" height="81" rx="6" fill="#8E3A80"/>
    </svg>
    """,
    "gender": """
    <svg width="160" height="140" viewBox="0 0 180 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="65" cy="35" r="15" fill="#D4DF38"/>
        <path d="M40 60 H90 L85 135 H45 Z" fill="#D4DF38"/>
        <circle cx="115" cy="35" r="15" fill="#8E3A80"/>
        <path d="M95 60 H135 L150 115 H130 V135 H100 V115 H85 Z" fill="#8E3A80"/>
    </svg>
    """,
    "rings": """
    <svg width="150" height="140" viewBox="0 0 160 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="68" cy="92" r="32" fill="none" stroke="#D4DF38" stroke-width="7"/>
        <circle cx="92" cy="92" r="32" fill="none" stroke="#8E3A80" stroke-width="7"/>
        <path d="M80 55 C80 55 62 42 62 30 C62 22 70 18 76 24 L80 28 L84 24 C90 18 98 22 98 30 C98 42 80 55 80 55 Z" fill="none" stroke="#8E3A80" stroke-width="6"/>
    </svg>
    """,
    "couple": """
    <svg width="170" height="140" viewBox="0 0 180 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="55" cy="28" r="16" fill="#D4DF38"/>
        <rect x="35" y="50" width="40" height="85" rx="5" fill="#D4DF38"/>
        <circle cx="125" cy="28" r="16" fill="#8E3A80"/>
        <path d="M105 50 H145 L162 115 H88 Z" fill="#8E3A80"/>
        <rect x="112" y="115" width="10" height="22" fill="#8E3A80"/>
        <rect x="128" y="115" width="10" height="22" fill="#8E3A80"/>
    </svg>
    """,
    "symbols": """
    <svg width="160" height="140" viewBox="0 0 180 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="72" cy="82" r="28" fill="none" stroke="#D4DF38" stroke-width="10"/>
        <path d="M72 110 V142 M56 128 H88" stroke="#D4DF38" stroke-width="10"/>
        <circle cx="98" cy="55" r="28" fill="none" stroke="#8E3A80" stroke-width="10"/>
        <path d="M126 55 H155 M140 38 L158 55 L140 72" fill="none" stroke="#8E3A80" stroke-width="10"/>
    </svg>
    """,
    "protection": """
    <svg width="160" height="140" viewBox="0 0 180 150" xmlns="http://www.w3.org/2000/svg">
        <rect x="35" y="35" width="36" height="90" rx="18" fill="none" stroke="#D4DF38" stroke-width="8"/>
        <text x="80" y="110" font-size="72" font-weight="bold" fill="#8E3A80">?</text>
        <rect x="115" y="35" width="38" height="90" rx="19" fill="#8E3A80"/>
    </svg>
    """,
    "symptoms": """
    <svg width="160" height="140" viewBox="0 0 180 150" xmlns="http://www.w3.org/2000/svg">
        <circle cx="90" cy="78" r="55" fill="none" stroke="#D4DF38" stroke-width="8" stroke-dasharray="60 25"/>
        <circle cx="90" cy="24" r="15" fill="#8E3A80"/>
        <path d="M40 50 H140 V68 H108 V135 H72 V68 H40 Z" fill="#8E3A80"/>
        <circle cx="62" cy="58" r="9" fill="#FFFFFF" stroke="#8E3A80" stroke-width="4"/>
        <circle cx="98" cy="78" r="9" fill="#FFFFFF" stroke="#8E3A80" stroke-width="4"/>
        <circle cx="78" cy="112" r="9" fill="#FFFFFF" stroke="#8E3A80" stroke-width="4"/>
    </svg>
    """,
}

# -----------------------------------------------------------------------------
# QUESTIONNAIRE DEFINITION
# -----------------------------------------------------------------------------
QUESTIONS = [
    {
        "id": "step_1",
        "step_label": "Step 1 of 12",
        "question": "What is your Age?",
        "svg": "age",
        "options": [
            "Less than 19",
            "19-34",
            "35-55",
            "More than 55",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_2",
        "step_label": "Step 2 of 12",
        "question": "What is your Gender?",
        "svg": "gender",
        "options": [
            "Female",
            "Male",
            "Transgender",
            "Non-binary / Gender diverse",
            "Prefer not to say",
        ],
    },
    {
        "id": "step_3",
        "step_label": "Step 3 of 12",
        "question": "Are you married?",
        "svg": "rings",
        "options": [
            "Yes",
            "No",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_3_1",
        "step_label": "Step 3.1 of 12",
        "question": "If married, What is your current status?",
        "svg": "couple",
        "condition": lambda ans: ans.get("step_3") == "Yes",
        "options": [
            "Living with spouse",
            "Living with a non-marital person",
            "Separated",
            "Do not wish to answer",
            "Not applicable",
        ],
    },
    {
        "id": "step_3_2",
        "step_label": "Step 3.2 of 12",
        "question": "Have you had sex with anyone outside your marital relationship?",
        "svg": "symbols",
        "condition": lambda ans: ans.get("step_3") == "Yes",
        "options": [
            "Yes",
            "No",
        ],
    },
    {
        "id": "step_4",
        "step_label": "Step 4 of 12",
        "question": "When was your last sexual encounter?",
        "svg": "symbols",
        "options": [
            "Less than a week",
            "More than a week less than a month",
            "More than a month",
            "Never",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_5",
        "step_label": "Step 5 of 12",
        "question": "How do you know / meet your sexual partner?",
        "svg": "symbols",
        "condition": lambda ans: ans.get("step_4") != "Never",
        "options": [
            "Spouse / Long-term regular partner",
            "Acquaintance",
            "Internet dating site/mobile app",
            "Facebook/Whatsapp friends",
            "Commercial Sex worker",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_6",
        "step_label": "Step 6 of 12",
        "question": "Did you or your partner use any protective method (e.g., condoms)?",
        "svg": "protection",
        "condition": lambda ans: ans.get("step_4") != "Never",
        "options": [
            "Yes, consistently",
            "Sometimes / Inconsistently",
            "No",
        ],
    },
    {
        "id": "step_7",
        "step_label": "Step 7 of 12",
        "question": "Have you had multiple sexual partners in the past 6 months?",
        "svg": "couple",
        "condition": lambda ans: ans.get("step_4") != "Never",
        "options": [
            "Yes",
            "No",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_8",
        "step_label": "Step 8 of 12",
        "question": "Did you have any symptoms in your genital / anal region / mouth / skin?",
        "svg": "symptoms",
        "options": [
            "Yes (sores, ulcers, unusual discharge, burning, or rash)",
            "No",
        ],
    },
    {
        "id": "step_9",
        "step_label": "Step 9 of 12",
        "question": "Has your sexual partner recently experienced any STI symptoms or been diagnosed with an STI/HIV?",
        "svg": "symptoms",
        "options": [
            "Yes",
            "No",
            "Not sure / Do not know",
        ],
    },
    {
        "id": "step_10",
        "step_label": "Step 10 of 12",
        "question": "Have you ever shared needles, syringes, or injecting equipment?",
        "svg": "shield",
        "options": [
            "Yes",
            "No",
            "Do not wish to answer",
        ],
    },
    {
        "id": "step_11",
        "step_label": "Step 11 of 12",
        "question": "Have you ever been tested for HIV or other STIs previously?",
        "svg": "shield",
        "options": [
            "Yes, within the last 6 months",
            "Yes, more than 6 months ago",
            "Never tested",
        ],
    },
    {
        "id": "step_12",
        "step_label": "Step 12 of 12",
        "question": "Would you like guidance on free confidential testing centres near you?",
        "svg": "shield",
        "options": [
            "Yes",
            "No",
        ],
    },
]

# -----------------------------------------------------------------------------
# SESSION STATE MANAGEMENT
# -----------------------------------------------------------------------------
if "page" not in st.session_state:
    st.session_state.page = "intro"
if "q_idx" not in st.session_state:
    st.session_state.q_idx = 0
if "answers" not in st.session_state:
    st.session_state.answers = {}


def get_active_questions():
    active = []
    for q in QUESTIONS:
        cond = q.get("condition")
        if cond is None or cond(st.session_state.answers):
            active.append(q)
    return active


def reset_assessment():
    st.session_state.page = "intro"
    st.session_state.q_idx = 0
    st.session_state.answers = {}


# -----------------------------------------------------------------------------
# HEADER COMPONENT
# -----------------------------------------------------------------------------
def render_header():
    st.markdown(
        """
        <div class="top-banner">
            <div class="banner-title-row">
                <span class="yellow-check">✔</span>
                <h1 class="banner-title">STI Self-Assessment</h1>
            </div>
            <div class="banner-subtitle">
                Know your risk. Take charge of your sexual health.
            </div>
        </div>
        """,
        unsafe_allow_html=True,
    )


# -----------------------------------------------------------------------------
# RISK EVALUATION ENGINE
# -----------------------------------------------------------------------------
def evaluate_risk(ans):
    score = 0
    reasons = []

    if str(ans.get("step_8", "")).startswith("Yes"):
        score += 4
        reasons.append(
            "Reported symptoms in the genital, anal, oral, or skin region require clinical evaluation."
        )

    if ans.get("step_9") == "Yes":
        score += 3
        reasons.append(
            "Partner history of STI symptoms or diagnosis increases exposure risk."
        )

    if ans.get("step_10") == "Yes":
        score += 4
        reasons.append(
            "Sharing needles or injecting equipment carries a high risk for blood-borne infections (HIV, Hepatitis B/C)."
        )

    if ans.get("step_6") in ["No", "Sometimes / Inconsistently"]:
        score += 2
        reasons.append(
            "Inconsistent or non-use of protective barriers (condoms) increases STI transmission risk."
        )

    if ans.get("step_3_2") == "Yes" or ans.get("step_7") == "Yes":
        score += 2
        reasons.append(
            "Having multiple or concurrent sexual partners increases the likelihood of exposure."
        )

    if ans.get("step_5") in [
        "Commercial Sex worker",
        "Internet dating site/mobile app",
        "Facebook/Whatsapp friends",
    ]:
        score += 1
        reasons.append(
            "Engaging with partners whose STI status may be unknown increases potential risk."
        )

    if ans.get("step_9") == "Not sure / Do not know":
        score += 1

    if score >= 4:
        level = "Higher Risk – Clinical Evaluation & Testing Recommended"
        color = "#C62828"
        bg = "#FFEBEE"
    elif score >= 2:
        level = "Moderate Risk – Routine STI/HIV Screening Advised"
        color = "#EF6C00"
        bg = "#FFF3E0"
    else:
        level = "Lower Reported Risk – Continue Preventive Practices"
        color = "#2E7D32"
        bg = "#E8F5E9"

    return level, color, bg, reasons


# -----------------------------------------------------------------------------
# APP PRESENTATION
# -----------------------------------------------------------------------------
render_header()

if st.session_state.page == "intro":
    st.markdown('<div class="assessment-card">', unsafe_allow_html=True)
    col1, col2 = st.columns([1.3, 1])

    with col1:
        st.markdown(
            """
            <h2 style="color:#8E3A80; font-size:28px; margin-bottom:8px;">
                TAKE A MOMENT TO UNDERSTAND YOUR STI RISK
            </h2>
            <p style="font-size:19px; color:#222; line-height:1.6; margin-top:16px;">
                Answer a few simple questions about your sexual health. This self-assessment
                will help you understand possible risk factors and identify practical next steps.
            </p>
            <p style="font-size:20px; color:#111; font-weight:500; margin-top:20px;">
                Are you ready?
            </p>
            """,
            unsafe_allow_html=True,
        )

        btn_col1, btn_col2 = st.columns([1, 1.5])
        with btn_col1:
            if st.button("Start Self-Assessment"):
                st.session_state.page = "instructions"
                st.rerun()

        st.markdown(
            """
            <div class="footer-note">
                *Note: The data collected as part of this assessment will not be used to
                determine your identity and the data is not stored anywhere.
            </div>
            <div class="initiative-badge">
                🎗️ Public Health Initiative | Confidential STI & HIV Risk Assessment
            </div>
            """,
            unsafe_allow_html=True,
        )

    with col2:
        st.markdown(
            f'<div style="text-align:center; padding-top:20px;">{SVGS["shield"]}</div>',
            unsafe_allow_html=True,
        )
    st.markdown("</div>", unsafe_allow_html=True)

elif st.session_state.page == "instructions":
    st.markdown(
        '<div class="step-bar"><div class="step-tab">Assessment Instructions</div></div>',
        unsafe_allow_html=True,
    )
    st.markdown('<div class="assessment-card">', unsafe_allow_html=True)
    st.markdown(
        """
        <h3 style="color:#8E3A80; margin-top:0;">Before You Begin</h3>
        <ul style="font-size:18px; line-height:1.8; color:#222;">
            <li><strong>Confidential & Anonymous:</strong> Your responses are processed only in your current browser session and are never saved or shared.</li>
            <li><strong>Non-Judgmental Guidance:</strong> Every question is asked solely to tailor health and testing information to your situation. You may select <em>"Do not wish to answer"</em> on sensitive demographic items.</li>
            <li><strong>Not a Medical Diagnosis:</strong> This tool estimates potential risk factors. Only a laboratory test or healthcare provider can confirm an STI or HIV diagnosis.</li>
        </ul>
        """,
        unsafe_allow_html=True,
    )

    c1, c2, c3 = st.columns([1, 1, 2])
    with c1:
        if st.button("← Back"):
            st.session_state.page = "intro"
            st.rerun()
    with c2:
        if st.button("Proceed to Questions →"):
            st.session_state.page = "quiz"
            st.session_state.q_idx = 0
            st.rerun()
    st.markdown("</div>", unsafe_allow_html=True)

elif st.session_state.page == "quiz":
    active_questions = get_active_questions()

    if st.session_state.q_idx >= len(active_questions):
        st.session_state.q_idx = len(active_questions) - 1

    current_q = active_questions[st.session_state.q_idx]
    q_id = current_q["id"]

    st.markdown(
        f'<div class="step-bar"><div class="step-tab">{current_q["step_label"]}</div></div>',
        unsafe_allow_html=True,
    )

    st.markdown('<div class="assessment-card">', unsafe_allow_html=True)
    left_col, right_col = st.columns([1.05, 1])

    with left_col:
        svg_markup = SVGS.get(current_q["svg"], SVGS["shield"])
        st.markdown(
            f"""
            <div class="question-box">
                <div>{svg_markup}</div>
                <div class="question-text">{current_q["question"]}</div>
            </div>
            """,
            unsafe_allow_html=True,
        )

    with right_col:
        st.write("")
        saved_val = st.session_state.answers.get(q_id, current_q["options"][0])
        default_idx = (
            current_q["options"].index(saved_val)
            if saved_val in current_q["options"]
            else 0
        )

        selected = st.radio(
            label=current_q["question"],
            options=current_q["options"],
            index=default_idx,
            key=f"radio_{q_id}",
            label_visibility="collapsed",
        )
        st.session_state.answers[q_id] = selected

    st.markdown("</div>", unsafe_allow_html=True)

    st.write("")
    foot_left, foot_center, foot_right = st.columns([1.2, 2, 0.5])

    with foot_left:
        st.markdown(
            '<div class="initiative-badge">🎗️ Confidential Public Health Assessment</div>',
            unsafe_allow_html=True,
        )

    with foot_center:
        b1, b2, b3 = st.columns(3)
        with b1:
            if st.button("Cancel"):
                reset_assessment()
                st.rerun()
        with b2:
            if st.session_state.q_idx > 0:
                if st.button("← Previous"):
                    st.session_state.q_idx -= 1
                    st.rerun()
        with b3:
            is_last = st.session_state.q_idx == len(active_questions) - 1
            next_label = "Finish ✔" if is_last else "Next →"
            if st.button(next_label):
                if is_last:
                    st.session_state.page = "results"
                else:
                    st.session_state.q_idx += 1
                st.rerun()

elif st.session_state.page == "results":
    st.markdown(
        '<div class="step-bar"><div class="step-tab">Assessment Summary & Next Steps</div></div>',
        unsafe_allow_html=True,
    )
    st.markdown('<div class="assessment-card">', unsafe_allow_html=True)

    level, color, bg, reasons = evaluate_risk(st.session_state.answers)

    st.markdown(
        f"""
        <div style="background-color:{bg}; border-left:8px solid {color}; padding:20px; border-radius:6px; margin-bottom:24px;">
            <h2 style="color:{color}; margin:0; font-size:26px;">{level}</h2>
            <p style="margin-top:8px; font-size:17px; color:#222;">
                <strong>Important Disclaimer:</strong> This self-assessment identifies potential risk factors
                and is <strong>not a medical diagnosis</strong>. Only a healthcare professional and laboratory
                testing can confirm an STI or HIV status.
            </p>
        </div>
        """,
        unsafe_allow_html=True,
    )

    col_a, col_b = st.columns(2)

    with col_a:
        st.markdown("### Key Considerations from Your Responses")
        if reasons:
            for r in reasons:
                st.markdown(f"- {r}")
        else:
            st.markdown(
                "- You did not report high-risk exposures or current physical symptoms."
            )
            st.markdown(
                "- Routine periodic screening is still a healthy habit for all sexually active individuals."
            )

        if st.session_state.answers.get("step_4") == "Less than a week":
            st.info(
                "**Time-Sensitive Note (PEP & Window Period):** Because your most recent encounter was within the last week, if you believe you were exposed to HIV within the last **72 hours**, speak to a doctor immediately about **Post-Exposure Prophylaxis (PEP)**. Also note that some tests have a 'window period' before an infection can be detected."
            )

    with col_b:
        st.markdown("### Practical Next Steps & Referrals")
        st.markdown(
            """
            1. **Confidential Testing:** Visit an Integrated Counselling and Testing Centre (**ICTC**) or Designated STI/RTI Clinic (**DSRC** / *Suraksha Clinic*) at your nearest Government Medical College or District Hospital for free, confidential testing.
            2. **National Toll-Free Helpline:** Call **1097** (National AIDS Control Organisation Helpline) for 24/7 confidential counselling and clinic locations.
            3. **Prevention Strategies:** Consistent use of external or internal condoms, routine partner testing, and discussing **PrEP** (Pre-Exposure Prophylaxis) with a clinician help keep you protected.
            """
        )

    st.write("")
    r1, r2, r3 = st.columns([1, 1, 2])
    with r1:
        if st.button("← Review Answers"):
            st.session_state.page = "quiz"
            st.rerun()
    with r2:
        if st.button("Start New Assessment"):
            reset_assessment()
            st.rerun()

    st.markdown("</div>", unsafe_allow_html=True)
```
